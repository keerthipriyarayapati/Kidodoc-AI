```python
import pandas as pd

train_df = pd.read_csv("../datasets/Training.csv")
test_df = pd.read_csv("../datasets/Testing.csv")

print(train_df.head())
```

```python
# Remove unnecessary column
train_df = train_df.drop(columns=["Unnamed: 133"])

# Check dataset shape
print("Training Shape:", train_df.shape)

# Check missing values
print(train_df.isnull().sum().sum())
```

```python
# Show all diseases
print(train_df["prognosis"].unique())

# Number of diseases
print("Total Diseases:", train_df["prognosis"].nunique())
```

```python
selected_diseases = [
    'Allergy',
    'GERD',
    'Drug Reaction',
    'Gastroenteritis',
    'Bronchial Asthma',
    'Migraine',
    'Malaria',
    'Chicken pox',
    'Dengue',
    'Typhoid',
    'Common Cold',
    'Pneumonia',
    'Urinary tract infection',
    'Acne',
    'Impetigo'
]

filtered_df = train_df[train_df['prognosis'].isin(selected_diseases)]

print(filtered_df['prognosis'].unique())
print("Total Selected Diseases:", filtered_df['prognosis'].nunique())
print("Filtered Shape:", filtered_df.shape)
```

```python
# Features
X = filtered_df.drop("prognosis", axis=1)

# Labels
y = filtered_df["prognosis"]

print("Features Shape:", X.shape)
print("Labels Shape:", y.shape)
```

```python
#Encode Labels
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

print(y_encoded[:10])
```

```python
#Train the data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
#First ml model-Random Forest
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()

model.fit(X_train, y_train)

print("Model Trained Successfully")
```

```python
#Test model Accuracy
from sklearn.metrics import accuracy_score
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
#Test real prediction
sample_prediction = model.predict([X.iloc[0]])
predicted_disease = encoder.inverse_transform(sample_prediction)

print("Predicted Disease:", predicted_disease[0])
```

```python
#Save the model
import pickle
pickle.dump(model, open("../backend/model.pkl", "wb"))

# Save encoder
pickle.dump(encoder, open("../backend/encoder.pkl", "wb"))

print("Model and Encoder Saved")
```
